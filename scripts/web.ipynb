```python
import requests
```

```python

BASE_URL = "http://wiener.lan"
USERNAME = "paperBot"

# 1️⃣ GET strony głównej (opcjonalne, np. dla CSRF tokena)
# r = requests.get(BASE_URL + "/")
# print("GET / status:", r.status_code)
# print(r.text)  # jeśli chcesz zobaczyć formularz

# 2️⃣ POST login
headers = {
    "Content-Type": "application/x-www-form-urlencoded",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "Origin": BASE_URL,
    "Referer": BASE_URL  + "/",
    "User-Agent": "Mozilla/5.0 (X11; Ubuntu; Linux x86_64; rv:147.0) Gecko/20100101 Firefox/147.0",
}

data = {
    "username": USERNAME
}

# używamy sesji, żeby automatycznie przechowywała cookies
session = requests.Session()
post_response = session.post(BASE_URL , data=data)
print("POST / status:", post_response.status_code)
print("Session cookie:", session.cookies.get_dict())

# 3️⃣ GET strony chronionej z cookie
protected_response = session.get(BASE_URL, headers={
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "User-Agent": headers["User-Agent"]
})

print("GET protected status:", protected_response.status_code)
print("--- Protected page content ---")
print(protected_response.text)
```
